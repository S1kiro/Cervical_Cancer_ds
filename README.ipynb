The hogh risk of existing cervical cancer in woman is caused by these factors:
1. *AGE* About 50% of cervical cancer diagnoses occur in women ages 35 - 54, and about 20% occur in women over 65 years of age. It's so rarely to meet cervical cancer in women under 20 years.
2. *SOCIOECONOMIC AND ETHNIC FACTORS* Due to the poor economical and social status of a woman, the percantage of existing a cervical cancer extremely increased. By this, it remains much more prevalent in African-Americans -- whose death rates are twice as high as Caucasian women.
3. *HIGH SEXUAL ACTIVITY* A huge history of multiple sexual partners, sexual intercourse at age 17 years or younger increase the risk of cervical cancer existing. A woman who has never been sexually active has a very low risk for developing cervical cancer.
4. *FAMILY HISTORY* Women have a higher risk of cervical cancer if they have a first-degree relative (mother, sister) who has had cervical cancer.
5. *USE OF ORAL CONTRACEPTIVES* Women who take birth control pills for more than 5 - 10 years appear to have a much higher risk HPV infection (up to four times higher) than those who do not use OCs.
6. *HAVING MANY CHILDREN* Studies indicate that having many children increases the risk for developing cervical cancer, particularly in women infected with HPV.
7. *SMOKING* Smoking is associated with a higher risk for precancerous changes (dysplasia) in the cervix and for progression to invasive cervical cancer, especially for women infected with HPV.
8. *IMMUNOSUPPRESSION* Women with weak immune systems, (such as those with HIV / AIDS), are more susceptible to acquiring HPV. Immunocompromised patients are also at higher risk for having cervical precancer develop rapidly into invasive cancer.
9. *DIETHYLSTILBESTROL* (DES) From 1938 - 1971, diethylstilbestrol (DES), an estrogen-related drug, was widely prescribed to pregnant women to help prevent miscarriages. The daughters of these women face a higher risk for cervical cancer. DES is no longer prsecribed.
